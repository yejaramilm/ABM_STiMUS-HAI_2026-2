```python
import pandas as pd

a = pd.read_csv("Results\\verification_modified.csv", skiprows=6)
b = pd.read_csv("Results\\verification_original.csv", skiprows=6)
```

```python
a = a.sort_values("[run number]").reset_index(drop=True)
a
```

```python
b = b.sort_values("[run number]").reset_index(drop=True)
b
```

```python
common = [c for c in a.columns if c in b.columns]
diff = (a[common] != b[common]).any()

print("Corridas:", len(a), len(b))
print("Columnas distintas:", list(diff[diff].index))
```
